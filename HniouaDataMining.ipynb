1.3. Prétraitement des Données

**Encoder la colonne 'Attack' pour en faire notre variable cible multiclasse**

```python
from sklearn.preprocessing import LabelEncoder

# Encoder la colonne 'Attack' pour en faire notre variable cible multiclasse
# Nous allons créer un nouveau LabelEncoder pour la variable cible y.
le_multiclass_target = LabelEncoder()
y = le_multiclass_target.fit_transform(df['Attack'])
```

**Définition de X (features) et y (variable cible) pour la classification multiclasse**

Pour l'objectif de classification multiclasse des attaques réseau, la variable cible `y` doit être le type d'attaque ('Attack') encodé numériquement, et les features `X` seront toutes les autres colonnes pertinentes, en excluant la colonne binaire 'Label' et la colonne originale 'Attack'.

```python
# Les features X: toutes les colonnes numériques sauf 'Label' et l'originale 'Attack'.
# Note: Le 'Label' binaire n'est pas utilisé comme feature pour prédire le type spécifique d'attaque.
X = df.drop(['Label', 'Attack'], axis=1)

print("Dimensions des features (X) pour la classification multiclasse:", X.shape)
print("Dimensions de la variable cible (y) pour la classification multiclasse:", y.shape)
print("\nClasses encodées pour la variable cible 'Attack':")
# Afficher le mapping des classes encodées pour comprendre la correspondance
for i, class_name in enumerate(le_multiclass_target.classes_):
    print(f"  {class_name}: {i}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd

# --- Normalisation AVANT le split ---
scaler = StandardScaler()
X_normalized = scaler.fit_transform(X)   # Normalisation de toutes les features

X_normalized = pd.DataFrame(X_normalized, columns=X.columns)  # Rétablir les noms de colonnes
X_normalized
```

* Division du jeu de données avec les nouveaux X et y

Maintenant que X et y sont correctement définis pour la tâche de classification multiclasse, nous devons refaire la division en ensembles d'entraînement et de test pour que `X_train`, `X_test`, `y_train` et `y_test` reflètent ces nouvelles définitions. Nous utiliserons toujours la stratification pour maintenir la proportion des classes d'attaque dans les deux ensembles.

```python
# --- Split des données (avec y stratifié pour conserver les proportions multiclasses) ---
X_train, X_test, y_train, y_test = train_test_split(
    X_normalized, y, test_size=0.20, random_state=42, stratify=y
)

# --- Vérification des dimensions ---
print("Dimensions de X_train :", X_train.shape)
print("Dimensions de X_test :", X_test.shape)
print("Dimensions de y_train :", y_train.shape)
print("Dimensions de y_test :", y_test.shape)

# --- Vérifier la distribution des classes ---
print("\nDistribution des classes dans y_train :")
display(pd.Series(y_train).value_counts(normalize=True))

print("\nDistribution des classes dans y_test :")
display(pd.Series(y_test).value_counts(normalize=True))

x_train_df = pd.DataFrame(X_train)
x_train_df
```

# Partie 2 : Modélisation et Entraînement

## Conception et Compilation du Modèle LSTM (Partie Classifieur)

Concevoir l'architecture du modèle LSTM pour la classification multiclasse, incluant des couches LSTM et Dropout, se terminant par une couche Dense avec activation softmax. Compiler le modèle avec une fonction de perte, un optimiseur et une métrique appropriés.


La première étape consiste à importer les modules nécessaires de TensorFlow/Keras pour construire et compiler le modèle LSTM, ainsi que pour encoder la variable cible.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dropout, Dense
from tensorflow.keras.utils import to_categorical
```

```python
num_classes = len(le_multiclass_target.classes_)
num_features = X_train_sm.shape[1]

# Reshape X_train_sm and X_test to 3D for LSTM (samples, timesteps, features)
X_train_sm_reshaped = X_train_sm.to_numpy().reshape(X_train_sm.shape[0], 1, num_features)
X_test_reshaped = X_test.to_numpy().reshape(X_test.shape[0], 1, num_features)

# One-hot encode y_train_sm and y_test
y_train_sm_onehot = to_categorical(y_train_sm, num_classes=num_classes)
y_test_onehot = to_categorical(y_test, num_classes=num_classes)

print(f"Number of classes: {num_classes}")
print(f"Number of features: {num_features}")
print(f"X_train_sm_reshaped shape: {X_train_sm_reshaped.shape}")
print(f"X_test_reshaped shape: {X_test_reshaped.shape}")
print(f"y_train_sm_onehot shape: {y_train_sm_onehot.shape}")
print(f"y_test_onehot shape: {y_test_onehot.shape}")
```

**Fonction de perte (loss function) :**
Pour un problème de classification multi-classes, nous utilisons la fonction categorical_crossentropy.


**Justification :** Cette fonction est adaptée lorsque la sortie du modèle est un vecteur de probabilités (obtenu avec softmax) et que chaque exemple appartient à une seule classe parmi plusieurs. Elle mesure la distance entre la distribution prédite par le modèle et la distribution réelle des classes, permettant ainsi au modèle de s’ajuster efficacement pour maximiser la probabilité de la classe correcte.

Construire le modèle LSTM en ajoutant successivement des couches LSTM avec le nombre d’unités spécifié, des couches Dropout pour la régularisation, et une couche Dense finale avec activation softmax pour la classification multi-classes, puis le compiler avec la fonction de perte, l’optimiseur et les métriques appropriés, et afficher le résumé du modèle.

```python
model = Sequential([
    LSTM(100, input_shape=(1, num_features), return_sequences=True),  # First LSTM layer
    Dropout(0.2),
    LSTM(100, return_sequences=False),  # Second LSTM layer
    Dropout(0.2),
    Dense(num_classes, activation='softmax')  # Output layer
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("LSTM Model Architecture:")
model.summary()
```

Entraîner le modèle LSTM conçu et compilé en utilisant les données d’entraînement suréchantillonnées, puis l’évaluer sur les données de test.

```python
history = model.fit(X_train_sm_reshaped, y_train_sm_onehot, epochs=10, batch_size=32, validation_data=(X_test_reshaped, y_test_onehot), verbose=1)

print("LSTM Model Training Complete.")
```

```python
import matplotlib.pyplot as plt

# Historique de l'entraînement
# history.history contient les listes : 'loss', 'val_loss', 'accuracy', 'val_accuracy'

# Tracer la loss
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(history.history['loss'], label='Loss entraînement')
plt.plot(history.history['val_loss'], label='Loss validation')
plt.title("Courbe de loss")
plt.xlabel("Époque")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

# Tracer l'accuracy
plt.subplot(1,2,2)
plt.plot(history.history['accuracy'], label='Accuracy entraînement')
plt.plot(history.history['val_accuracy'], label='Accuracy validation')
plt.title("Courbe d'accuracy")
plt.xlabel("Époque")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

## Extraction des Caractéristiques avec le Modèle LSTM Entraîné

Créer un nouveau modèle Keras à partir du modèle LSTM entraîné pour l'extraction de caractéristiques et générer les caractéristiques (embeddings) à partir des ensembles d'entraînement et de test.


Nouveau modèle Keras qui servira d’extracteur de features en supprimant la dernière couche de sortie du modèle LSTM préalablement entraîné.
Ce modèle retournera les features (embeddings) générées par les couches LSTM avant la couche de classification finale.

```python
feature_extractor = Sequential(model.layers[:-1])

print("Feature Extractor Model Architecture:")
feature_extractor.summary()
```

Générer les features (embeddings) pour les jeux de données d’entraînement et de test en utilisant le modèle feature_extractor. Cela implique d’appeler la méthode predict sur les données d’entraînement et de test préalablement reshaped.

```python
X_train_features = feature_extractor.predict(X_train_sm_reshaped)
X_test_features = feature_extractor.predict(X_test_reshaped)

print("Dimensions of X_train_features:", X_train_features.shape)
print("Dimensions of X_test_features:", X_test_features.shape)
```

```python
# Installer si besoin
# !pip install xgboost

import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Initialisation du modèle
xgb_model = xgb.XGBClassifier(
    n_estimators=100,      # nombre d'arbres
    max_depth=6,           # profondeur maximale de chaque arbre
    learning_rate=0.1,     # taux d'apprentissage
    subsample=0.8,         # échantillonnage aléatoire des données
    colsample_bytree=0.8,  # échantillonnage des features
    random_state=42,
    n_jobs=-1              # utilise tous les cœurs CPU
)
```

```python
xgb_model.fit(X_train_features, y_train_sm_onehot)
```

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt
import numpy as np

# --- Liste des labels dans l’ordre ---
class_names = [
    "Analysis", "Benign", "Bot", "Fuzzers", "Generic", "Infilteration",
    "Reconnaissance", "Shellcode", "Theft", "Backdoor", "Bruteforce",
    "DDoS", "DoS", "Injection", "MITM", "Ransomware", "Scanning", "XSS"
]

n_classes = len(class_names)

# --- y_test est 1D : on le binarise (one-hot) ---
y_test_onehot = label_binarize(y_test_classes, classes=list(range(n_classes)))

# --- Prédiction probabiliste ---
y_score = xgb_model.predict_proba(X_test_features)

# --- Courbe ROC multiclasses ---
plt.figure(figsize=(12, 10))

for i in range(n_classes):
    fpr, tpr, _ = roc_curve(y_test_onehot[:, i], y_score[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, lw=2, label=f'{class_names[i]} (AUC = {roc_auc:.2f})')

# Ligne diagonale
plt.plot([0, 1], [0, 1], 'k--', lw=2)

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Multi-classes - XGBoost", fontsize=16)
plt.legend(loc="lower right", fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
